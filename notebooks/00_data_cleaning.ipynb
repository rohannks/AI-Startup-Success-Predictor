# AI Startup Success Predictor

## Data Cleaning

This notebook loads and cleans the raw startup datasets.

The project uses three datasets:

- AI Startup Funding Dataset
- India AI Startups Dataset
- Global Unicorn Dataset

The cleaned datasets will be used for dataset integration, EDA, feature engineering, and machine learning.

```python
#IMPORT REQUIRED LIBRARIES


# Pandas is used for loading and manipulating tabular data.
import pandas as pd

# NumPy is used for numerical operations.
import numpy as np

# Path is used to create reliable file paths.
from pathlib import Path

# Display library is used to show DataFrames neatly in Jupyter.
from IPython.display import display

print("Libraries imported successfully. ✅")
```

Define paths

Why: This keeps our file paths organized and avoids hardcoding Windows paths.

```python
# Define the folder containing the raw datasets
RAW_DATA_PATH = Path("../data/raw")

# Define the folder where cleaned datasets will be saved
PROCESSED_DATA_PATH = Path("../data/processed")

# Create the processed folder if it does not exist
PROCESSED_DATA_PATH.mkdir(
    parents=True,
    exist_ok=True
)

print("Raw data folder:", RAW_DATA_PATH)
print("Processed data folder:", PROCESSED_DATA_PATH)
```

Find the datasets

Why: Instead of manually typing the long funding filename, we let Python find the files from the data/raw folder.

```python
# Find the AI startup funding dataset
funding_files = list(
    RAW_DATA_PATH.glob("ai_startup_funding_database*.csv")
)

# Find the India AI startup dataset
india_files = list(
    RAW_DATA_PATH.glob("india_ai_startups_complete.csv")
)

# Find the global unicorn dataset
unicorn_files = list(
    RAW_DATA_PATH.glob("List of Unicorns in the World.csv")
)

print("Funding files found:", funding_files)
print("India AI files found:", india_files)
print("Unicorn files found:", unicorn_files)
```

Verify the files

Why: This confirms that all three required datasets are present before we try to load them.

```python
# Make sure exactly one funding dataset was found
assert len(funding_files) == 1, (
    f"Expected 1 funding dataset, found {len(funding_files)}"
)

# Make sure exactly one India AI dataset was found
assert len(india_files) == 1, (
    f"Expected 1 India AI dataset, found {len(india_files)}"
)

# Make sure exactly one unicorn dataset was found
assert len(unicorn_files) == 1, (
    f"Expected 1 unicorn dataset, found {len(unicorn_files)}"
)

print("All three raw datasets found successfully. ✅")
```

Load all 3 datasets

Purpose: Load the three raw CSV files.

```python
# Load the three datasets

funding_df = pd.read_csv(funding_files[0])

india_df = pd.read_csv(
    india_files[0],
    engine="python",
    on_bad_lines="warn"
)

unicorn_df = pd.read_csv(unicorn_files[0])

print("All datasets loaded successfully. ✅")
```

Check dataset sizes

Purpose: Confirm the datasets loaded correctly.

```python
# Check dataset shapes

print("Funding:", funding_df.shape)
print("India AI:", india_df.shape)
print("Unicorn:", unicorn_df.shape)
```

Check columns

Purpose: See the original columns before cleaning.

```python
# Check column names

print("Funding columns:")
print(funding_df.columns.tolist())

print("\nIndia AI columns:")
print(india_df.columns.tolist())

print("\nUnicorn columns:")
print(unicorn_df.columns.tolist())
```

Clean column names

Purpose: Remove unwanted spaces from column names.

```python
# Remove extra spaces from column names

funding_df.columns = funding_df.columns.str.strip()
india_df.columns = india_df.columns.str.strip()
unicorn_df.columns = unicorn_df.columns.str.strip()

print("Column names cleaned. ✅")
```

Check missing values

Purpose: Find missing data. We will decide how to handle it later during feature engineering.

```python
# Check missing values

print("Funding missing values:")
print(funding_df.isnull().sum())

print("\nIndia AI missing values:")
print(india_df.isnull().sum())

print("\nUnicorn missing values:")
print(unicorn_df.isnull().sum())
```

Check duplicates

Purpose: Find duplicate rows.

```python
# Check duplicate rows

print("Funding duplicates:", funding_df.duplicated().sum())
print("India AI duplicates:", india_df.duplicated().sum())
print("Unicorn duplicates:", unicorn_df.duplicated().sum())
```

Remove duplicates

Purpose: Remove exact duplicate records if any exist.

```python
# Remove duplicate rows

funding_df = funding_df.drop_duplicates()
india_df = india_df.drop_duplicates()
unicorn_df = unicorn_df.drop_duplicates()

print("Duplicates removed. ✅")
```

Fix India AI data types

Purpose: Convert important numeric columns into numbers.

```python
# Convert India AI numeric columns

india_numeric = [
    "founded_year",
    "total_funding_usd",
    "latest_round_amt_usd",
    "latest_round_year",
    "valuation_usd"
]

for col in india_numeric:
    india_df[col] = pd.to_numeric(
        india_df[col],
        errors="coerce"
    )

print("India AI numeric columns cleaned. ✅")
```

Clean Unicorn valuation

Purpose: Convert $225 into numeric 225.

```python
# Convert unicorn valuation into numeric values

unicorn_df["Valuation ($B)"] = (
    unicorn_df["Valuation ($B)"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

unicorn_df["Valuation ($B)"] = pd.to_numeric(
    unicorn_df["Valuation ($B)"],
    errors="coerce"
)

print("Unicorn valuation cleaned. ✅")
```

Final validation

Purpose: Confirm everything is correct before saving.

```python
# Final dataset validation

print("========== FINAL DATASET VALIDATION ==========")

print("Funding Dataset:", funding_df.shape)
print("India AI Dataset:", india_df.shape)
print("Unicorn Dataset:", unicorn_df.shape)

print("\nFunding columns:", len(funding_df.columns))
print("India AI columns:", len(india_df.columns))
print("Unicorn columns:", len(unicorn_df.columns))
```

Save cleaned datasets

Purpose: Save the cleaned data into data/processed/.

```python
# Save cleaned datasets

funding_df.to_csv(
    "../data/processed/funding_cleaned.csv",
    index=False
)

india_df.to_csv(
    "../data/processed/india_ai_startups_cleaned.csv",
    index=False
)

unicorn_df.to_csv(
    "../data/processed/unicorn_cleaned.csv",
    index=False
)

print("All cleaned datasets saved successfully. ✅")
```
